```python
import numpy as np
import pandas as pd
import pickle
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
car = pd.read_csv('auto-mpg.csv',header=0)
```

```python
print(car)
```

```python
car.isnull()
```

```python
car['horsepower'] = pd.to_numeric(car['horsepower'], errors='coerce')
car = car.dropna()
```

```python
car = car.drop(['car_name'], axis=1)
```

```python
car = pd.get_dummies(car, columns=['origin'], drop_first=True, dtype=float)
```

```python
Y = car['mpg']
X = car.drop(['mpg'], axis=1)
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=0.3, random_state=0)
```

```python
lr = LinearRegression()
lr.fit(X_train, np.log(Y_train))
```

```python
pred = np.exp(lr.predict(X_test))
```

```python
print("테스트 R²:", round(r2_score(Y_test, pred), 3))
print("평균 오차(MAE):", round(mean_absolute_error(Y_test, pred), 2))
```

```python
with open('car.pkl', 'wb') as f:
    pickle.dump({'model': lr, 'columns': list(X.columns)}, f)
```

```python
import pickle
import numpy as np
import pandas as pd

with open('car.pkl', 'rb') as f:
    saved = pickle.load(f)
model = saved['model']
columns = saved['columns']

# 예시: 실린더 4, 배기량 150, 마력 100, 무게 2500, 가속 15, 연식 80, 출신 미국(1)
new = pd.DataFrame([[4, 150, 100, 2500, 15, 80, 0, 0]], columns=columns)
print("예상 연비:", round(float(np.exp(model.predict(new))[0]), 1), "mpg")
```
